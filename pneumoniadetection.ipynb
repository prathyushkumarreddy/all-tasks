```python
import os
import cv2
import matplotlib.pyplot as plt

DIRECTORY = "/content/pneumonia-detection-dataset/chest_xray/train"
categories = ["PNEUMONIA", "NORMAL"]
IMAGE_SIZE = 128

data = []

for category in categories:
    folder = os.path.join(DIRECTORY, category)
    label = categories.index(category)
    print(category, label)


    if not os.path.exists(folder):
        print(f"Warning: Folder not found: {folder}. Skipping.")
        continue

    for img in os.listdir(folder):
        img_path = os.path.join(folder, img)
        img_arr = cv2.imread(img_path)

        if img_arr is not None:
            img_arr = cv2.resize(img_arr, (IMAGE_SIZE, IMAGE_SIZE))
            data.append([img_arr, label])
```

```python
import random

sample_index = random.randint(0, len(data) - 1)
sample_image, sample_label = data[sample_index]

category_name = categories[sample_label]

plt.figure(figsize=(6, 6))
plt.imshow(sample_image)
plt.title(f"Label: {category_name}")
plt.axis('off')
plt.show()
```

```python
pneumonia_image = None
normal_image = None

for img_arr, label in data:
    if label == 0 and pneumonia_image is None:  # PNEUMONIA
        pneumonia_image = img_arr
    elif label == 1 and normal_image is None:  # NORMAL
        normal_image = img_arr

    if pneumonia_image is not None and normal_image is not None:
        break

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(pneumonia_image)
plt.title('PNEUMONIA')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(normal_image)
plt.title('NORMAL')
plt.axis('off')

plt.tight_layout()
plt.show()
```

TRAINING AND TESTING THE MODEL

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
import tensorflow as tf

```

```python
from keras.models import Sequential
from keras.layers import Conv2D
from keras.layers import MaxPooling2D
from keras.layers import Dropout
```

```python
from keras.layers import Flatten,Dense
```

```python
from tensorflow.keras.models import Sequential
from keras.layers import Conv2D
from keras.layers import MaxPooling2D
from keras.layers import Dropout
from keras.layers import Flatten,Dense

cnn = Sequential()
cnn.add(Conv2D(filters=64,kernel_size=(3,3),padding="same",activation="relu",input_shape=(128,128,3)))
cnn.add(MaxPooling2D(2,2))
cnn.add(Conv2D(filters=32,kernel_size=(3,3),padding="same",activation="relu"))
cnn.add(MaxPooling2D(2,2))
cnn.add(Dropout(0.2))
cnn.add(Conv2D(filters=32,kernel_size=(2,2),padding="same",activation="relu"))
cnn.add(MaxPooling2D(2,2))
cnn.add(Conv2D(filters=32,kernel_size=(2,2),padding="same",activation="relu"))
cnn.add(MaxPooling2D(2,2))
cnn.add(Dropout(0.2))
cnn.add(Conv2D(filters=128,kernel_size=(1,1),padding="same",activation="relu"))
cnn.add(MaxPooling2D(2,2))
cnn.add(Conv2D(filters=128,kernel_size=(1,1),padding="same",activation="relu"))
cnn.add(MaxPooling2D(2,2))
cnn.add(Dropout(0.2))

cnn.add(Flatten())
cnn.add(Dense(512,activation="relu"))
cnn.add(Dense(300,activation="relu"))
cnn.add(Dense(200,activation="relu"))
cnn.add(Dense(4,activation="softmax"))
```

```python
cnn.summary()
```

```python
cnn.compile(optimizer="adam",
            loss="sparse_categorical_crossentropy",
            metrics=["accuracy"])
```

```python
history = cnn.fit(x_train, y_train, epochs=8, validation_split=0.1)
```

```python
loss,accuracy = cnn.evaluate(x_test,y_test)
print("Test accuracy:",accuracy)
```

```python
pred=cnn.predict(x_test)
```

```python
pred
```

```python
index=[np.argmax(i) for i in pred]
```

```python
new_dataframe = pd.DataFrame({"actual":y_test,"predict":index})
print(new_dataframe)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
print(classification_report(y_test,index))
```

```python
cn=confusion_matrix(y_test,index)
```

```python
import seaborn as sns
sns.heatmap(cn,annot=True,square=True,fmt=".0f")
plt.show()
```

```python
acc=history.history['accuracy']
val_acc=history.history['val_accuracy']
epochs=range(len(acc))
fig = plt.figure(figsize=(13,6))
plt.plot(epochs,acc,'r',label="Training accuracy")
plt.plot(epochs,val_acc,'g',label="Validation accuracy")
plt.legend(loc="upper left")
plt.show()
```

```python
loss=history.history['loss']
val_loss=history.history['val_loss']
epochs=range(len(loss))
fig = plt.figure(figsize=(13,6))
plt.plot(epochs,loss,'r',label="Training loss")
plt.plot(epochs,val_loss,'g',label="Validation loss")
plt.legend(loc="upper left")
plt.show()
```
